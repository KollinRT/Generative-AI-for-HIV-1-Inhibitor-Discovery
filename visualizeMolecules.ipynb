```python
from rdkit import Chem
from rdkit.Chem import Draw

# List of SMILES strings
mols = [
    "CN(Br)OOC(N=N[CH]SCl)(P[N+]([NH])OOOCON)C(C)([C]O)[SH]=CC=O",
    "[O-]C1CCCON=[PH]2[CH]OC(N2)S1",
    "[CH]PC=CON[C@@H](NNC1(C=C=CC=C=CCC)[CH]OO1)OC(O)SF"
]

# Convert SMILES to RDKit molecule objects
molecules = [Chem.MolFromSmiles(mol) for mol in mols]

# Generate image grid with SMILES underneath
img = Draw.MolsToGridImage(
    molecules, 
    molsPerRow=3, 
    subImgSize=(300, 300), 
    legends=mols
)

# Save the image
img.save("molecules_grid.png")
```

```python
from rdkit import Chem
from rdkit.Chem import Draw
from PIL import Image

# List of SMILES strings
mols = [
    "CN(Br)OOC(N=N[CH]SCl)(P[N+]([NH])OOOCON)C(C)([C]O)[SH]=CC=O",
    "[O-]C1CCCON=[PH]2[CH]OC(N2)S1",
    "[CH]PC=CON[C@@H](NNC1(C=C=CC=C=CCC)[CH]OO1)OC(O)SF",
    "C=C1C=[C][C@@H]2C=CC(=O)C1OOC(C(=S)CF)=S=NC2",
    "CC1=CC(C(=CC(O)Cl)CN=C([CH]N)[CH]C=O)N(O[O-])C=C1",
    "C[CH]CC=[SH]C1=C(CNCC2NC(F)=CO[PH]3(C=CSCO2)O[N+](C=O)[C@H]3[O-])OSO1",
]

# Convert SMILES to RDKit molecule objects
molecules = [Chem.MolFromSmiles(mol) for mol in mols]

# Generate image grid with SMILES underneath (ensuring it's a PIL image)
img = Draw.MolsToGridImage(
    molecules, 
    molsPerRow=3, 
    subImgSize=(600, 600), 
    legends=mols,
    returnPNG=False
)

# Save the image
img.save("molecules_grid.png")

print("Image saved as molecules_grid.png")
```
