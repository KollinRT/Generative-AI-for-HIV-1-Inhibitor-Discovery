```python
import os
import csv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------- FINETUNE-SPECIFIC RENAME MAP (unique names) ----------
RENAME = {
    # BART base
    "selfies_BART_finetune_skip_base": "bart_adam_nosched_p5_ep100",
    # Lamb cosine (d30) ep=500
    "selfies_BART_finetune_model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th": "lamb_cos_d30_lr1e-6_p10_ep500",
    # Lamb cosine variants
    "selfies_BART_finetune_model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_1": "lamb_cos_d30_lr1e-6_p10_ep500",
    "selfies_BART_finetune_model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_2": "lamb_cos_d10_lr1e-6_p100_ep500",
    "selfies_BART_finetune_model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_3_long": "lamb_cos_d10_lr1e-5_p250_ep5000",
    "selfies_BART_finetune_model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_4": "lamb_cos_d10_lr5e-4_p250_ep100_bs32",
    "selfies_BART_finetune_model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_5": "lamb_cos_d10_lr3e-5_p250_ep100_bs32",
    # Lamb linear/cos small runs 1..4
    "selfies_BART_finetune_model_small_lamb_earlyS_1": "lamb_lin_d10_lr1e-6_p20_ep500",
    "selfies_BART_finetune_model_small_lamb_earlyS_2": "lamb_lin_d10_lr1e-6_p15_ep100",
    "selfies_BART_finetune_model_small_lamb_earlyS_3": "lamb_cos_d10_lr1e-7_p15_ep100",
    "selfies_BART_finetune_model_small_lamb_earlyS_4": "lamb_lin_d10_lr1e-6_p15_ep50_wu05",
    # AdamW series
    "selfies_BART_finetune_model_small_adamw_earlyS_1": "adamw_lin_d10_lr1e-6_p20_ep500",
    "selfies_BART_finetune_model_small_adamw_earlyS_2": "adamw_lin_d10_lr1e-6_p20_ep50",
    "selfies_BART_finetune_model_small_adamw_earlyS_3": "adamw_lin_d10_lr1e-4_p20_ep100",
    "selfies_BART_finetune_model_small_adamw_earlyS_4": "adamw_cos_d10_lr1e-4_p20_ep100",
    "selfies_BART_finetune_model_small_adamw_earlyS_5": "adamw_lin_d10_lr1e-4_p20_ep100_wu05",
    "selfies_BART_finetune_model_small_adamw_earlyS_6": "adamw_cos_d10_lr1e-4_p20_ep100_wu05",
    "selfies_BART_finetune_model_small_adamw_earlyS_6_long": "adamw_cos_d10_lr1e-4_p20_ep200_wu05",
    "selfies_BART_finetune_model_small_adamw_earlyS_6_long_3x": "adamw_cos_d10_lr1e-4_p20_ep300_wu05",
    # Baseline tuned
    "selfies_BART_finetune_model_baseline_tuned": "adamw_cos_d10_lr3e-4_p15_ep100_bs64",
}

# If your folders don’t have the exact "selfies_BART_finetune_" prefix,
# add additional keys for plain names too:
RENAME.update(
    {
        "model_baseline_tuned": "adamw_cos_d10_lr3e-4_p15_ep100_bs64",
        "model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_5": "lamb_cos_d10_lr3e-5_p250_ep100_bs32",
        "model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_4": "lamb_cos_d10_lr5e-4_p250_ep100_bs32",
        "model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_3_long": "lamb_cos_d10_lr1e-5_p250_ep5000",
        "model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_2": "lamb_cos_d10_lr1e-6_p100_ep500",
        "model_small_lamb_earlyS_extradropout_highLR_lowThresh_or7th_1": "lamb_cos_d30_lr1e-6_p10_ep500",
        "model_small_lamb_earlyS_4": "lamb_lin_d10_lr1e-6_p15_ep50_wu05",
        "model_small_lamb_earlyS_3": "lamb_cos_d10_lr1e-7_p15_ep100",
        "model_small_lamb_earlyS_2": "lamb_lin_d10_lr1e-6_p15_ep100",
        "model_small_lamb_earlyS_1": "lamb_lin_d10_lr1e-6_p20_ep500",
        "model_small_adamw_earlyS_6_long_3x": "adamw_cos_d10_lr1e-4_p20_ep300_wu05",
        "model_small_adamw_earlyS_6_long": "adamw_cos_d10_lr1e-4_p20_ep200_wu05",
        "model_small_adamw_earlyS_6": "adamw_cos_d10_lr1e-4_p20_ep100_wu05",
        "model_small_adamw_earlyS_5": "adamw_lin_d10_lr1e-4_p20_ep100_wu05",
        "model_small_adamw_earlyS_4": "adamw_cos_d10_lr1e-4_p20_ep100",
        "model_small_adamw_earlyS_3": "adamw_lin_d10_lr1e-4_p20_ep100",
        "model_small_adamw_earlyS_2": "adamw_lin_d10_lr1e-6_p20_ep50",
        "model_small_adamw_earlyS_1": "adamw_lin_d10_lr1e-6_p20_ep500",
    }
)

RENAME_KEYS_BY_LEN = sorted(RENAME.keys(), key=len, reverse=True)


def display_name(run_folder: str) -> str:
    """Return pretty name via longest-first substring mapping."""
    for key in RENAME_KEYS_BY_LEN:
        if key in run_folder:
            return RENAME[key]
    # fallback: keep original folder
    return run_folder


def slugify(text: str) -> str:
    text = text.strip()
    text = re.sub(r"[ \t/]+", "-", text)
    text = re.sub(r"[^A-Za-z0-9._-]+", "_", text)
    return text


def pick_x(df: pd.DataFrame):
    for c in ["global_step", "step", "iteration", "it"]:
        if c in df.columns:
            x = pd.to_numeric(df[c], errors="coerce")
            if x.notna().sum() > 1 and x.nunique(dropna=True) > 1:
                return ("Global Step", x)
    if "epoch" in df.columns:
        x = pd.to_numeric(df["epoch"], errors="coerce")
        if x.notna().sum() > 1 and x.nunique(dropna=True) > 1:
            return ("Epoch", x)
    return ("Step (index)", pd.Series(np.arange(len(df)), index=df.index))


def ensure_numeric(df: pd.DataFrame, cols):
    out = df.copy()
    for c in cols:
        if c in out.columns:
            out[c] = pd.to_numeric(out[c], errors="coerce")
    return out


# -------- PATHS (update runs_root) --------
runs_root = "/home/kollin/Desktop/ExploreThesis/WIP_Thesis/runs"
# only finetune dirs:
dirs = [
    d
    for d in os.listdir(runs_root)
    if os.path.isdir(os.path.join(runs_root, d)) and "finetune" in d
]

output_dir = "./plots/finetune/renamed_files"
os.makedirs(output_dir, exist_ok=True)

mapping_log = []
ROLL = 0  # set >0 to smooth

# to avoid overwriting if names collide unexpectedly
used_basenames = set()

for run in sorted(dirs):
    csv_path = os.path.join(runs_root, run, "finetune_loss.csv")
    if not os.path.exists(csv_path):
        print(f"Skipping (missing CSV): {csv_path}")
        continue
    try:
        df = pd.read_csv(csv_path)
        df = ensure_numeric(
            df,
            [
                "epoch",
                "global_step",
                "step",
                "iteration",
                "train_loss",
                "val_loss",
                "validation_loss",
                "eval_loss",
            ],
        )

        if "val_loss" not in df.columns:
            if "validation_loss" in df.columns:
                df["val_loss"] = df["validation_loss"]
            elif "eval_loss" in df.columns:
                df["val_loss"] = df["eval_loss"]
            else:
                raise ValueError(
                    "No validation loss column found (val_loss/validation_loss/eval_loss)."
                )
        if "train_loss" not in df.columns:
            raise ValueError("Missing 'train_loss'.")

        x_label, x_series = pick_x(df)
        df = df.assign(_x=x_series).sort_values("_x")

        y_tr = df["train_loss"]
        y_va = df["val_loss"]
        if ROLL and len(df) >= ROLL:
            y_tr = y_tr.rolling(ROLL, min_periods=1).mean()
            y_va = y_va.rolling(ROLL, min_periods=1).mean()

        pretty = display_name(run)
        safe = slugify(pretty)

        # collision-proof output name
        base_png = f"{safe}_loss.png"
        if base_png in used_basenames:
            i = 2
            while f"{safe}-v{i}_loss.png" in used_basenames:
                i += 1
            base_png = f"{safe}-v{i}_loss.png"
        used_basenames.add(base_png)

        plt.figure(figsize=(10, 6))
        plt.plot(df["_x"], y_tr, label="Train Loss", alpha=0.85)
        plt.plot(df["_x"], y_va, label="Validation Loss", alpha=0.85)
        plt.xlabel(x_label)
        plt.ylabel("Loss")
        plt.title(f"Finetune Loss — {pretty}")
        plt.legend()
        plt.grid(True)

        plot_path = os.path.join(output_dir, base_png)
        plt.savefig(plot_path, bbox_inches="tight", dpi=150)
        plt.close()
        print(f"Saved plot: {plot_path}")

        mapping_log.append((run, pretty, x_label, plot_path))

    except Exception as e:
        print(f"Error processing {csv_path}: {e}")

# write mapping log
mapping_csv = os.path.join(output_dir, "mapping_original_to_short.csv")
with open(mapping_csv, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["original_folder", "short_name_used", "x_axis_used", "plot_path"])
    writer.writerows(mapping_log)

print(f"Saved mapping log: {mapping_csv}")
```
